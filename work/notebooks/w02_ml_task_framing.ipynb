## 3. Success metric

*Precision@K (specifically Precision@20 and Precision@50) is the metric I'd
defend — not plain accuracy. The output is a ranked review queue with
limited reviewer capacity, so what matters is: of the top K pages flagged,
how many are actually declining? That matches exactly how the output gets used.*

```python
print("Success metric: Precision@20 and Precision@50")
print("'Good' = beats the hand-written baseline rule on these metrics")
```
