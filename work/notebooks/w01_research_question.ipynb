## 3. Quick look at the data (2-3 real numbers)

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = df.shape[0]
declining_rate = df["trend_direction"].str.lower().eq("down").mean()
avg_impressions_declining = df[df["trend_direction"].str.lower() == "down"]["impressions_90d"].mean()

print(f"Total pages: {total_pages:,}")
print(f"Share currently declining: {declining_rate:.1%}")
print(f"Avg impressions_90d for declining pages: {avg_impressions_declining:.0f}")
```
